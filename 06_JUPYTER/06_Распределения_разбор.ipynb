# Тема 6. Разбор: два одинаковых средних

Демонстрация для занятия. Ячейки можно менять; исходные CSV остаются без изменений.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
DATA = Path('../05_ДАННЫЕ_СТУДЕНТАМ')
assert DATA.exists(), 'Запускайте ноутбук из папки 06_JUPYTER рядом с учебными файлами.'
```

```python
df = pd.read_csv(DATA / 'campus_distributions_06.csv')
df['co2_ppm'] = pd.to_numeric(df['co2_ppm'], errors='raise')
assert df['observation_id'].is_unique
print(df.groupby('room_code')['co2_ppm'].agg(['count','mean','median','min','max']))
```

## Распределения с общими интервалами

Почему нельзя сравнивать гистограммы, если в каждой программе автоматически выбраны свои границы?

```python
fig, ax = plt.subplots(figsize=(8, 4))
bins = range(400, 2801, 200)
for room in sorted(df.room_code.unique()):
    df.loc[df.room_code == room, 'co2_ppm'].plot.hist(ax=ax, bins=bins, alpha=.5, label=room)
ax.set(xlabel='CO₂, ppm', ylabel='Количество наблюдений', title='Два распределения при одинаковом среднем')
ax.legend();plt.tight_layout();plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6,3))
df.boxplot(column='co2_ppm', by='room_code', ax=ax)
plt.suptitle('');ax.set(xlabel='Аудитория',ylabel='CO₂, ppm');plt.tight_layout();plt.show()
```

## Условное исключение высоких наблюдений

Нельзя менять исходный CSV. Ниже сравниваются **две разные выборки** для отдельного расчёта.

```python
b = df[df.room_code == 'B-201']['co2_ppm']
print('Все:',len(b),round(b.mean(),1),'ppm')
print('Без показаний 2500:',len(b[b!=2500]),round(b[b!=2500].mean(),1),'ppm')
assert len(df) == 80
```
