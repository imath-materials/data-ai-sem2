# Тема 7. Разбор: знак связи меняется после разделения на группы

Цель — отличить наблюдаемую корреляцию от объяснения её причин.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
DATA = Path('../05_ДАННЫЕ_СТУДЕНТАМ')
assert DATA.exists(), 'Запускайте ноутбук из папки 06_JUPYTER рядом с учебными файлами.'
```

```python
df = pd.read_csv(DATA / 'campus_relationships_07.csv')
assert len(df)==108
print('Всего наблюдений:',len(df))
print('Общая корреляция:',round(df['people_count'].corr(df['co2_ppm']),3))
print('По режимам:')
for name,part in df.groupby('vent_mode'):
    print(name,len(part),round(part['people_count'].corr(part['co2_ppm']),3))
```

```python
fig, ax = plt.subplots(figsize=(8,5))
for mode,part in df.groupby('vent_mode'):
    ax.scatter(part['people_count'],part['co2_ppm'],alpha=.72,label=mode)
ax.set(xlabel='Людей на сессии',ylabel='CO₂, ppm',title='Общий график с разбиением по режиму')
ax.legend();plt.tight_layout();plt.show()
```

## Условия не одинаковы

Одновременно меняются режим вентиляции, вместимость и помещение. Поэтому нельзя утверждать, что только режим является причиной различия CO₂.

```python
print(df.groupby('vent_mode')[['people_count','room_capacity','co2_ppm']].agg(['min','max','mean']).round(1))
```

## Как должен звучать итог

**Наблюдение:** общая и внутригрупповые корреляции имеют разные знаки. **Ограничение:** группы различаются по нескольким факторам. **План:** получить сопоставимые наблюдения, уточнить период и условия измерений.
